### **Step 2 — Select only houses for sale**
The dataset is filtered to isolate rows representing houses for sale. This ensures the model focuses specifically on predicting selling prices for residential houses.

```python
house_df = df[
    (df["category"] == "Houses For Sale") &
    (df["type"] == "for_sale")
].copy()

print("Remaining rows:", len(house_df))
```

### **Step 3 — Convert the `price` column to numbers**
Since price values are stored as string text, the numeric digits representing prices in LKR are extracted, and any rows with invalid or missing pricing are dropped.

```python
house_df["price_lkr"] = pd.to_numeric(
    house_df["price"]
    .str.extract(r"Rs\s*([\d,]+(?:\.\d+)?)")[0]
    .str.replace(",", "", regex=False),
    errors="coerce"
)

house_df[["price", "price_lkr"]].head()
```

```python
house_df = house_df.dropna(subset=["price_lkr"])
```

### **Step 4 — Extract bedrooms, bathrooms and property sizes**
The dictionary-like structure inside the `properties` column is parsed to extract key numeric features such as bedrooms, bathrooms, and house size in square feet.

```python
house_df.head()
```

```python
import ast
import re

# Safely convert dictionary-like text into dictionaries
def parse_dict(value):
    try:
        result = ast.literal_eval(value)
        return result if isinstance(result, dict) else {}
    except (ValueError, SyntaxError, TypeError):
        return {}

house_df["property_dict"] = house_df["properties"].apply(parse_dict)

# Extract numeric part from a value
def extract_number(value):
    if pd.isna(value):
        return np.nan

    match = re.search(r"\d[\d,]*(?:\.\d+)?", str(value))

    if match:
        return float(match.group().replace(",", ""))

    return np.nan

house_df["bedrooms"] = house_df["property_dict"].apply(
    lambda x: extract_number(x.get("Bedrooms"))
)

house_df["bathrooms"] = house_df["property_dict"].apply(
    lambda x: extract_number(x.get("Bathrooms"))
)

house_df["house_sqft"] = house_df["property_dict"].apply(
    lambda x: extract_number(x.get("House size"))
)
```

### **Converting Land Units to Perches**
Land sizes are processed by identifying and normalizing units—converting acreage values into perches so that all land areas are measured on a consistent scale.

```python
def convert_land_to_perches(value):
    number = extract_number(value)

    if pd.isna(number):
        return np.nan

    text = str(value).lower()

    if "acre" in text:
        return number * 160

    if "perch" in text:
        return number

    return np.nan

house_df["land_perches"] = house_df["property_dict"].apply(
    lambda x: convert_land_to_perches(x.get("Land size"))
)
```

### **Step 5 — Extract the district and location**
The dictionary in the `area` column is parsed to extract the broader `district` name, which is paired with the local `location` column.

```python
house_df["district"] = house_df["area"].apply(
    lambda x: parse_dict(x).get("name")
)

house_df[["district", "location"]].head()
```

### **Filtering Records with Reasonable Bounds**
Extreme outliers are removed by parsing the date column and defining conservative bounds for price, house size, and land size. Subsequently, rows with missing values in key features are dropped.

```python
# Parse dates before final cleaning
house_df["posted_date"] = pd.to_datetime(
    house_df["posted_date"],
    errors="coerce",
    utc=True
)

house_df = house_df[
    house_df["price_lkr"].between(1_000_000, 500_000_000) &
    house_df["house_sqft"].between(300, 20_000) &
    house_df["land_perches"].between(1, 500)
].copy()

house_df = house_df.dropna(
    subset=["posted_date", "district", "location",
            "bedrooms", "bathrooms"]
)

print("Remaining records:", len(house_df))
```

### **Step 8 — Identify duplicate and reposted advertisements**
The data is sorted chronologically, and duplicate advertisements posted by the same users with identical features are eliminated to avoid test set leakage.

```python
# Sort by advertisement date
house_df = house_df.sort_values("posted_date")

# Identify likely reposted properties
duplicate_columns = [
    "user",
    "ad_description",
    "properties",
    "location"
]

house_df = house_df.drop_duplicates(
    subset=duplicate_columns,
    keep="first"
)

print("Records after deduplication:", len(house_df))
```

```python
model_df = house_df[
    [
        "posted_date",
        "district",
        "location",
        "bedrooms",
        "bathrooms",
        "house_sqft",
        "land_perches",
        "price_lkr"
    ]
].copy()

display(model_df.head())
```

### **Chronological Train-Test Split**
The dataset is sorted chronologically by the advertisement date. The first 80% of the listings are allocated to the training set, and the final 20% are allocated to the test set.

```python
# Sort cleaned dataset chronologically
model_df = model_df.sort_values(
    "posted_date"
).reset_index(drop=True)

# Input features
feature_cols = [
    "district",
    "location",
    "bedrooms",
    "bathrooms",
    "house_sqft",
    "land_perches"
]

# 80% training, 20% testing
split_index = int(len(model_df) * 0.8)

train_df = model_df.iloc[:split_index].copy()
test_df = model_df.iloc[split_index:].copy()

# Features
X_train = train_df[feature_cols]
X_test = test_df[feature_cols]

# Target in millions of LKR
y_train = train_df["price_lkr"] / 1_000_000
y_test = test_df["price_lkr"] / 1_000_000

print("Training:", X_train.shape)
print("Testing:", X_test.shape)
```

### **Feature Preprocessing Pipeline Definition**
A pipeline is structured using a `ColumnTransformer` to handle preprocessing. Median imputation is mapped to numerical features, and most frequent imputation followed by One-Hot Encoding is mapped to categorical features.

```python
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder

numeric_features = [
    "bedrooms",
    "bathrooms",
    "house_sqft",
    "land_perches"
]

categorical_features = [
    "district",
    "location"
]

def make_preprocessor():
    numeric_pipeline = Pipeline([
        ("imputer", SimpleImputer(
            strategy="median"
        ))
    ])

    categorical_pipeline = Pipeline([
        ("imputer", SimpleImputer(
            strategy="most_frequent"
        )),
        ("encoder", OneHotEncoder(
            handle_unknown="ignore",
            min_frequency=80
        ))
    ])

    return ColumnTransformer([
        ("numeric", numeric_pipeline, numeric_features),
        ("categorical", categorical_pipeline, categorical_features)
    ])
```

### **Random Forest Regression Training and Prediction**
A Random Forest Regressor is integrated into a scikit-learn pipeline, trained on the prepared features, and used to generate price predictions on the held-out test set.

```python
X_train.head()
```

```python
from sklearn.ensemble import RandomForestRegressor

rf_pipeline = Pipeline([
    ("preprocessor", make_preprocessor()),
    ("model", RandomForestRegressor(
        n_estimators=60,
        max_depth=12,
        min_samples_leaf=4,
        max_features=0.65,
        random_state=42,
        n_jobs=4
    ))
])

rf_pipeline.fit(X_train, y_train)

rf_predictions = rf_pipeline.predict(X_test)
```

### **XGBoost Regression Training and Prediction**
An Extreme Gradient Boosting (XGBoost) Regressor is configured and fit on the training data, then evaluated on the test set.

```python
from xgboost import XGBRegressor

xgb_pipeline = Pipeline([
    ("preprocessor", make_preprocessor()),
    ("model", XGBRegressor(
        n_estimators=300,
        max_depth=5,
        learning_rate=0.05,
        subsample=0.85,
        colsample_bytree=0.85,
        reg_lambda=2.0,
        tree_method="hist",
        random_state=42,
        n_jobs=4
    ))
])

xgb_pipeline.fit(X_train, y_train)

xgb_predictions = xgb_pipeline.predict(X_test)
```

### **Baseline and Machine Learning Model Evaluation**
Mean Absolute Error (MAE), Root Mean Squared Error (RMSE), and R-squared (R²) scores are calculated to evaluate the predictive performance of Random Forest and XGBoost.

```python
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

def evaluate_model(name, y_true, predictions):
    return {
        "Model": name,
        "MAE": mean_absolute_error(
            y_true, predictions
        ),
        "RMSE": np.sqrt(
            mean_squared_error(y_true, predictions)
        ),
        "R2": r2_score(
            y_true, predictions
        )
    }

results = pd.DataFrame([
    evaluate_model(
        "Random Forest", y_test, rf_predictions
    ),
    evaluate_model(
        "XGBoost", y_test, xgb_predictions
    )
])

display(results.round(3))
```

### **Performance Comparison Against Median Baseline**
A naive baseline model that predicts the median price of the training set is evaluated. This provides a reference point to measure the improvement of the regression models.

```python
baseline_predictions = np.full(
    len(y_test),
    y_train.median()
)

baseline_result = evaluate_model(
    "Median Baseline",
    y_test,
    baseline_predictions
)

results = pd.concat(
    [pd.DataFrame([baseline_result]), results],
    ignore_index=True
)

display(results.round(3))
```

### **Actual vs Predicted Price Visualization**
Scatter plots showing actual versus predicted prices are generated for both the Random Forest and XGBoost models to visually assess fit across the price spectrum.

```python
fig, axes = plt.subplots(1, 2, figsize=(14, 5))

models = [
    ("Random Forest", rf_predictions),
    ("XGBoost", xgb_predictions)
]

for ax, (name, predictions) in zip(axes, models):
    ax.scatter(
        y_test,
        predictions,
        alpha=0.3,
        s=10
    )

    ax.plot(
        [0, 200],
        [0, 200],
        "r--",
        label="Perfect Prediction"
    )

    ax.set_xlim(0, 200)
    ax.set_ylim(0, 200)
    ax.set_xlabel("Actual Price (LKR Millions)")
    ax.set_ylabel("Predicted Price (LKR Millions)")
    ax.set_title(name)
    ax.legend()

plt.tight_layout()
plt.show()
```
